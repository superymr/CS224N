## 1.1：提取词表（官方分值 2 分）

实现 `distinct_words(corpus)`。输入是多篇已分词文档，返回去重后按 Python 默认规则排序的词列表 `corpus_words`，以及词表大小 `n_corpus_words`。大小写视为不同词；保留输入已有的边界标记，不做额外预处理。


```python
def distinct_words(corpus):
    """返回排序后的唯一词列表，以及词表大小。"""
    unique_words = set() #保存不重复的词表
    for sentence in corpus:
        for word in sentence:
            if word not in unique_words:
                unique_words.add(word)

    sorted_words = sorted(unique_words)
    num = len(sorted_words)
    return sorted_words, num
```

```python
# 自编检查：不是官方检查，也不是完整评测。
words, n_words = distinct_words(corpus)
expected_words = sorted([START_TOKEN, END_TOKEN, '我', '喜欢', '学习', '编程'])
assert words == expected_words, (words, expected_words)
assert n_words == len(expected_words)
assert distinct_words([]) == ([], 0)
assert distinct_words([['a', 'a', 'A']]) == (['A', 'a'], 2)
print('题目 1.1：检查通过')
```
